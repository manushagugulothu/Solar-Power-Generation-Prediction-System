```python
# Cell: Load and preview the data
df = pd.read_csv('/content/SolarPrediction.csv')

print("Shape (rows, columns):", df.shape)
df.head()
```

```python
# Check column names, data types, and missing values
df.info()
```

```python
# See min, max, average of each numeric column
df.describe()
```

```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Extract just the date part from 'Data' (ignore its dummy 12:00:00 AM)
df['DateOnly'] = df['Data'].str.split(' ').str[0]

# Combine the real date with the real time
df['DateTime'] = pd.to_datetime(df['DateOnly'] + ' ' + df['Time'], format='%m/%d/%Y %H:%M:%S')

# Extract useful time-based features
df['Hour'] = df['DateTime'].dt.hour
df['Minute'] = df['DateTime'].dt.minute
df['Month'] = df['DateTime'].dt.month
df['DayOfYear'] = df['DateTime'].dt.dayofyear

df[['DateTime', 'Hour', 'Minute', 'Month', 'DayOfYear']].head()
```

```python
# Convert sunrise/sunset times to proper time format
df['TimeSunRise'] = pd.to_datetime(df['TimeSunRise'], format='%H:%M:%S').dt.time
df['TimeSunSet'] = pd.to_datetime(df['TimeSunSet'], format='%H:%M:%S').dt.time

# Create an "IsDaytime" feature: 1 if current time is between sunrise and sunset, else 0
df['CurrentTime'] = df['DateTime'].dt.time
df['IsDaytime'] = df.apply(lambda row: 1 if row['TimeSunRise'] <= row['CurrentTime'] <= row['TimeSunSet'] else 0, axis=1)

df[['DateTime', 'TimeSunRise', 'TimeSunSet', 'IsDaytime']].head()
```

```python
# Sort by actual datetime, oldest to newest
df = df.sort_values('DateTime').reset_index(drop=True)
df[['DateTime']].head()
```

```python
# Features (inputs) the model will learn from
features = ['Hour', 'Month', 'DayOfYear', 'Temperature', 'Pressure',
            'Humidity', 'WindDirection(Degrees)', 'Speed', 'IsDaytime']

# Target (what we want to predict)
target = 'Radiation'

X = df[features]
y = df[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
X.head()
```

```python
from sklearn.model_selection import train_test_split

# 80% earliest data for training, 20% latest data for testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
import xgboost as xgb

# Create the model
model = xgb.XGBRegressor(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    random_state=42
)

# Train it
model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

predictions = model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print(f"RMSE: {rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,5))
plt.plot(y_test.values[:300], label='Actual Radiation', color='orange')
plt.plot(predictions[:300], label='Predicted Radiation', color='blue', alpha=0.7)
plt.xlabel('Time (sample index)')
plt.ylabel('Radiation (W/m²)')
plt.title('Actual vs Predicted Solar Radiation')
plt.legend()
plt.show()
```

```python
import numpy as np

# Cyclical encoding for Hour (so 23 and 0 are recognized as close)
df['Hour_sin'] = np.sin(2 * np.pi * df['Hour'] / 24)
df['Hour_cos'] = np.cos(2 * np.pi * df['Hour'] / 24)

# Cyclical encoding for DayOfYear (so Dec 31 and Jan 1 are recognized as close)
df['Day_sin'] = np.sin(2 * np.pi * df['DayOfYear'] / 365)
df['Day_cos'] = np.cos(2 * np.pi * df['DayOfYear'] / 365)

# Temperature x Humidity interaction (weather features often interact)
df['Temp_Humidity'] = df['Temperature'] * df['Humidity']

df[['Hour_sin', 'Hour_cos', 'Day_sin', 'Day_cos', 'Temp_Humidity']].head()
```

```python
# Updated feature list with the new engineered features
features = ['Hour_sin', 'Hour_cos', 'Day_sin', 'Day_cos', 'Month',
            'Temperature', 'Pressure', 'Humidity', 'WindDirection(Degrees)',
            'Speed', 'IsDaytime', 'Temp_Humidity']

target = 'Radiation'

X = df[features]
y = df[target]

# Re-split (same 80/20 time-based split)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
# Train the improved model
model = xgb.XGBRegressor(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    random_state=42
)
model.fit(X_train, y_train)

# Evaluate
predictions = model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print(f"New RMSE: {rmse:.2f}")
print(f"New R² Score: {r2:.4f}")
```

```python
# Previous readings (lag features) - what was radiation a few steps ago?
df['Radiation_lag1'] = df['Radiation'].shift(1)   # 5 minutes ago
df['Radiation_lag2'] = df['Radiation'].shift(2)   # 10 minutes ago
df['Radiation_lag3'] = df['Radiation'].shift(3)   # 15 minutes ago

# Rolling average - smoothed recent trend
df['Radiation_roll3'] = df['Radiation'].rolling(window=3).mean()

# These create some empty (NaN) values at the very start since there's no "before" data yet
df = df.dropna().reset_index(drop=True)

df[['Radiation', 'Radiation_lag1', 'Radiation_lag2', 'Radiation_lag3', 'Radiation_roll3']].head()
```

```python
from sklearn.model_selection import GridSearchCV

# Different combinations to try
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [4, 6, 8],
    'learning_rate': [0.05, 0.1, 0.2]
}

grid_search = GridSearchCV(
    estimator=xgb.XGBRegressor(random_state=42),
    param_grid=param_grid,
    cv=3,                     # splits training data 3 ways to test each combo
    scoring='r2',
    verbose=1
)

grid_search.fit(X_train, y_train)

print("Best parameters found:", grid_search.best_params_)
print("Best R² during search:", grid_search.best_score_)
```

```python
# Use the best model found by the search
best_model = grid_search.best_estimator_

predictions = best_model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print(f"Tuned Model RMSE: {rmse:.2f}")
print(f"Tuned Model R² Score: {r2:.4f}")
```

```python
import pandas as pd

importance = pd.DataFrame({
    'feature': features,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

print(importance)
```

```python
from sklearn.model_selection import TimeSeriesSplit

tscv = TimeSeriesSplit(n_splits=3)

grid_search = GridSearchCV(
    estimator=xgb.XGBRegressor(random_state=42),
    param_grid=param_grid,
    cv=tscv,              # respects time order now
    scoring='r2',
    verbose=1
)

grid_search.fit(X_train, y_train)

print("Best parameters found:", grid_search.best_params_)
print("Best R² during search:", grid_search.best_score_)

best_model = grid_search.best_estimator_
predictions = best_model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print(f"Tuned Model RMSE: {rmse:.2f}")
print(f"Tuned Model R² Score: {r2:.4f}")
```

```python
# Remove features that add no value (confirmed by importance scores)
final_features = ['Hour_sin', 'Hour_cos', 'Day_sin', 'Day_cos',
                   'Temperature', 'Pressure', 'Humidity',
                   'WindDirection(Degrees)', 'Speed', 'Temp_Humidity']

X = df[final_features]
y = df['Radiation']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

final_model = xgb.XGBRegressor(
    n_estimators=200,
    max_depth=8,
    learning_rate=0.2,
    random_state=42
)
final_model.fit(X_train, y_train)

predictions = final_model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print(f"Final Model RMSE: {rmse:.2f}")
print(f"Final Model R² Score: {r2:.4f}")
```
